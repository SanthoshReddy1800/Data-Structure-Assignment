# **Q1. How many movies are present in the dataset?**

```python
import pandas as pd

df = pd.read_csv("ToP 250 movies on imdb in 2026.csv")

print(len(df))
```

# **Q2. How many columns are present and what are their data types?**

```python
print(len(df.columns))
print(df.dtypes)
```

# **Q3. Which columns contain missing values? Show the count for each.**

```python
missing = df.isnull().sum()

print(missing[missing > 0])
```

# **Q4. Are there duplicate movie records? If yes, identify and remove them.**

```python
print(df.duplicated().sum())

df = df.drop_duplicates()

print(len(df))
```

# **Q5. What is the average IMDb rating of all movies?**

```python
print(df["averageRating"].mean())
```

# **Q6. What is the median IMDb rating?**

```python
print(df["averageRating"].median())
```

# **Q7. What is the average movie runtime in minutes?**

```python
print(df["runtimeMinutes"].mean())
```

# **Q8. Which five movies have the highest IMDb ratings?**

```python
top_movies = df.nlargest(5, "averageRating")

print(top_movies[["primaryTitle", "averageRating"]])
```

# **Q9. Which five movies have the highest number of votes?**

```python
top_votes = df.nlargest(5, "numVotes")

print(top_votes[["primaryTitle", "numVotes"]])
```

# **Q10. Which five movies have the highest worldwide gross?**

```python
top_gross = df.nlargest(5, "grossWorldwide")

print(top_gross[["primaryTitle", "grossWorldwide"]])
```

# **Q11. Which five movies have the highest budget?**

```python
top_budget = df.nlargest(5, "budget")

print(top_budget[["primaryTitle", "budget"]])
```

# **Q12. Which five movies have the highest Metascore?**

```python
top_metascore = df.nlargest(5, "metascore")

print(top_metascore[["primaryTitle", "metascore"]])
```

# **Q13. Which movie has the longest runtime?**

```python
movie = df.loc[df["runtimeMinutes"].idxmax()]

print(movie[["primaryTitle", "runtimeMinutes"]])
```

# **Q14. Which movie has the shortest runtime?**

```python
movie = df.loc[df["runtimeMinutes"].idxmin()]

print(movie[["primaryTitle", "runtimeMinutes"]])
```

# **Q15. How many movies were released in each year? Show the top 10 years.**

```python
movies_year = df.groupby("startYear").size()

print(movies_year.sort_values(ascending=False).head(10))
```

# **Q16. Calculate the average IMDb rating for each release year.**

```python
average_rating = df.groupby("startYear")["averageRating"].mean()

print(average_rating)
```

# **Q17. Find the relationship between the number of votes and IMDb rating. Create a scatter plot.**

```python
import matplotlib.pyplot as plt

print(df["numVotes"].corr(df["averageRating"]))

plt.scatter(df["numVotes"], df["averageRating"])

plt.xlabel("Number of Votes")
plt.ylabel("IMDb Rating")
plt.title("Votes vs IMDb Rating")

plt.show()
```

# **Q18. Does movie runtime appear to have any relationship with IMDb rating? Visualize it.**

```python
import matplotlib.pyplot as plt

print(df["runtimeMinutes"].corr(df["averageRating"]))

plt.scatter(df["runtimeMinutes"], df["averageRating"])

plt.xlabel("Runtime in Minutes")
plt.ylabel("IMDb Rating")
plt.title("Runtime vs IMDb Rating")

plt.show()
```

# **Q19. Find the top 10 movies based on the ratio**

grossWorldwide
budget

```python
ratio = df.dropna(
    subset=["budget", "grossWorldwide"]
).copy()

ratio = ratio[ratio["budget"] > 0]

ratio["ratio"] = ratio["grossWorldwide"] / ratio["budget"]

print(
    ratio.nlargest(10, "ratio")[
        ["primaryTitle", "budget", "grossWorldwide", "ratio"]
    ]
)
```
